# 1.1.1 Aggregated Transaction Table Creartion

```python
import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="root",                                               #Connecting to My SQL
    password="12345678"
)

cursor = connection.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS phonepe_db")      #Creating Database


cursor.execute("USE phonepe_db")

# Creating Aggregated Transaction Table     # Creating Aggregated Transaction Table     # Creating Aggregated Transaction Table

cursor.execute("""
CREATE TABLE IF NOT EXISTS aggregated_transaction (
    State VARCHAR(100),
    Year INT,
    Quarter_Num INT,
    Quarter VARCHAR(10),
    Category_Name VARCHAR(100),
    Count BIGINT,
    Amount DOUBLE
)
""")

print("Table created successfully!")


connection.close()
```

```python
import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="12345678"
)

cursor = connection.cursor()

cursor.execute("USE phonepe_db")

                                            # Renaming columns for better clarity in aggregated_transaction table
cursor.execute("""
ALTER TABLE aggregated_transaction
CHANGE COLUMN Count Transaction_count BIGINT,
CHANGE COLUMN Amount Transaction_amount DOUBLE
""")

                                                    # Fetch and display updated table structure
cursor.execute("DESCRIBE aggregated_transaction")

columns = cursor.fetchall()

print("\nUpdated Table Structure:\n")
for col in columns:
    print(f"Column: {col[0]} | Type: {col[1]}")

connection.close()
```
