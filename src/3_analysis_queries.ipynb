```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="12345678",
    database="phonepe_db"
)

                                                #Query to get top 10 states by transaction amount
query = """
SELECT 
    State, 
    SUM(Amount) AS Total_Amount,
    FORMAT(SUM(Amount)/10000000, 2) AS Amount_in_Crore
FROM aggregated_transaction
GROUP BY State
ORDER BY Total_Amount DESC
LIMIT 10;
"""

                                                #Execute and load into DataFrame
df = pd.read_sql(query, conn)


print(df)
conn.close()
```

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="12345678",
    database="phonepe_db"
)

                                                #Query to get state-wise list by transaction count
query = """
SELECT 
    State, 
    SUM(Count) AS Total_Count,
    SUM(Count)/10000000 AS Count_in_Crore
FROM aggregated_transaction
GROUP BY State
ORDER BY Total_Count DESC;
"""

                                                #Execute and load into DataFrame
df = pd.read_sql(query, conn)


print(df)
conn.close()
```
