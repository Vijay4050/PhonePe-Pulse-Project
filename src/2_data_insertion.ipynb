# 1.1.1 Aggregated Transaction
## Data Extraction and Cleaning

```python
import os
import json
import pandas as pd

path = r"D:/Viki/Guvi DS/Capstone Projects/PhonePe-Pulse-Project/pulse/data/aggregated/transaction/country/india/state"

data = []

# Looping through states, years, quarters, untill to JSON file which contains the required data
for state in os.listdir(path):
    state_path = os.path.join(path, state)

    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)

        for file in os.listdir(year_path):
            if file.endswith(".json"):
                file_path = os.path.join(year_path, file)

                quarter_num = int(file.split(".")[0])
                quarter = f"FY{quarter_num}"

                with open(file_path, 'r') as file:           # Loading JSON File
                    content = json.load(file)

                if 'data' in content and 'transactionData' in content['data']:

                    for item in content['data']['transactionData']:
                        category_name = item.get('name')

                        paymentInstruments = item.get('paymentInstruments', [])

                        if paymentInstruments:
                            transaction_count = paymentInstruments[0].get('count', 0)
                            transaction_amount = paymentInstruments[0].get('amount', 0)
                        else:
                            transaction_count = 0
                            transaction_amount = 0

                        
                        data.append({
                            "State": state,
                            "Year": int(year),
                            "Quarter_Num" : quarter_num,
                            "Quarter": f"FY{quarter_num}",            # Appending Each data
                            "Category_Name": category_name,
                            "Transaction_count": transaction_count,
                            "Transaction_amount": transaction_amount
                        })


df = pd.DataFrame(data)

print(df.head())            #To view the headings of the created Data Frame


df.to_csv("phonepe_transactions.csv", index=False)  

print("********** Data extraction completed and saved as CSV file **********")
```

# 1.2.1 Aggregated User
## Data Extraction and Cleaning

```python
import os
import json
import pandas as pd

# 🔹 Path to aggregated user data
path = r"D:/Viki/Guvi DS/Capstone Projects/PhonePe-Pulse-Project/pulse/data/aggregated/user/country/india/state"

data = []

# 🔹 Looping  through each folder to get the required json files
for state in os.listdir(path):
    state_path = os.path.join(path, state)
    
    
    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)
        
        
        for file in os.listdir(year_path):
            
            if file.endswith(".json"):
                file_path = os.path.join(year_path, file)
                
                with open(file_path, "r") as f:
                    content = json.load(f)
                    
                    try:
                        # Extract user data
                        users = content["data"]["usersByDevice"]
                        
                        for row in users:
                            data.append({
                                "State": state,
                                "Year": int(year),
                                "Quarter_Num": int(file.strip(".json")),
                                "Quarter": "Q" + file.strip(".json"),
                                "Brand": row["brand"],
                                "Registered_users": row["count"],
                                "Device_share": row["percentage"]   # NOTE: No app opens in dataset → using percentage
                            })
                    
                    except:
                        # Some files may not have usersByDevice
                        continue

df = pd.DataFrame(data)

df.to_csv("aggregated_user.csv", index=False)

print("Aggregated User CSV created successfully!")
print(df.head())
```

# 1.3.1 Aggregated insurance
## Data Extraction and Cleaning

```python
import os
import json
import pandas as pd

# Set your correct base path
path = r"D:/Viki/Guvi DS/Capstone Projects/PhonePe-Pulse-Project/pulse/data/aggregated/insurance/country/india/state"

data = []

# Loop through states
for state in os.listdir(path):
    state_path = os.path.join(path, state)

    # Loop through years
    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)

        # Loop through quarters
        for quarter_file in os.listdir(year_path):
            file_path = os.path.join(year_path, quarter_file)

            with open(file_path, "r") as file:
                content = json.load(file)

                # Extracting data from JSON files and appending to the list and saving it as a CSV file
                try:
                    for item in content["data"]["transactionData"]:
                        name = item["name"]

                        for payment in item["paymentInstruments"]:
                            insurance_count = payment["count"]
                            insurance_amount = payment["amount"]

                            data.append({
                                "State": state,
                                "Year": int(year),
                                "Quarter": int(quarter_file.strip(".json")),
                                "Insurance_Type": name,
                                "Insurance_Count": insurance_count,
                                "Insurance_Amount": insurance_amount
                            })

                except:
                    pass  

# Convert to DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("aggregated_insurance.csv", index=False)

print("✅ Aggregated Insurance CSV created successfully!")
print(df.head())
```

# 2.1.1 Map User
## Data Extraction and Cleaning

```python
import os
import json
import pandas as pd

path = r"D:/Viki/Guvi DS/Capstone Projects/PhonePe-Pulse-Project/pulse/data/map/user/hover/country/india/state"

data = []

for state in os.listdir(path):
    state_path = os.path.join(path, state)

    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)

        for quarter_file in os.listdir(year_path):
            file_path = os.path.join(year_path, quarter_file)

            with open(file_path, "r") as file:
                content = json.load(file)

                try:
                    users = content["data"]["hoverData"]

                    for district, details in users.items():
                        data.append({
                            "State": state,
                            "Year": int(year),
                            "Quarter": int(quarter_file.strip(".json")),
                            "District": district,
                            "Registered_Users": details["registeredUsers"],
                            "App_Opens": details["appOpens"]
                        })

                except:
                    pass

# Convert to DataFrame
df = pd.DataFrame(data)

df.to_csv("map_user.csv", index=False)

print("CSV file 'map_user.csv' created successfully!")
#----------------------------------------------------------------------------------------------------------------------

                                                # Display extracted data

print("Extracted Data Preview:")
print(df.head())
print("\nTotal Rows:", df.shape[0])


```
